Flexible Image Transport System (FITS) is a digital file format used for storage, transmission, and processing of data. Each FITS file is composed of two parts: a header and tables. The header provides all of the metadata (description of the data) for the image, and the tables are the raw data.

A SunPy map is an object which holds two-dimensional data (as a NumPy array) along with the accompanying metadata. So it not only carries the map but also its associated attributes such as observatory, instrument, exposure time, etc.

The wavelength of this image is 174 Å, so it is in the ultraviolet range.

The angular size of one pixel is 0.492 arcsec.

The spacecraft distance to the Sun is 44301351381 m and the pixel scale is 0.492 arcsec/pixel. Converting the angular size to radians gives 2.385e-6 radians, which then taking the product with the distance yields roughly 106km per pixel.
